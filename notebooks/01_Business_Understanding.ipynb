# 1. Business Understanding
**Machine Learning 2026/2027 · Group 06**

This project predicts whether a customer will buy again within 90 days of a
purchase. We follow CRISP-DM and start by defining the prediction problem,
the information available and the decisions the analysis should support.

## Business context
Everything & Then Some, Ltd. is a wholesale company with customers whose
purchasing frequency and order values vary considerably. Historical sales
can help the commercial team organise follow-up actions.

The model will estimate repeat-purchase behaviour. The brief does not supply
campaign costs, margins or the effect of contacting a customer, so we cannot
claim an increase in profit or a causal effect from these data alone.

## Prediction problem

| Element | Definition |
|---|---|
| Observation | One customer purchase occasion, identified by `ID`. |
| Customer | Identified by `customer_id`; one customer can have several observations. |
| Target | `repeat_purchase_90d`: 1 for another purchase within the following 90 days; 0 otherwise. |
| Prediction moment | At the current purchase, when its details and previous history are available. |
| Task | Supervised binary classification. |
| Use | Support commercial prioritisation; the exact contact policy remains a business decision. |

`purchase_date` is the date of the **current** purchase. The target concerns
a subsequent purchase, not the purchase that preceded it.

## Data supplied

| File | Observations | Contents |
|---|---:|---|
| `train.csv` | 6,534 | Purchase information and the target. |
| `test.csv` | 986 | Later purchases with hidden targets. |
| `clientes.csv` | 1,143 | Optional customer attributes, joined through `customer_id`. |
| `sample_submission.csv` | 986 | IDs and the required prediction format. |

Predictors describe the current purchase, previous purchase timing and value,
and frequency/spend in earlier 30-, 90-, 180- and 365-day windows. Customer
attributes provide additional context but should not simply identify a client.

Dates are anonymised. The brief guarantees their ordering and intervals;
they should not be used to make claims about real historical events.

## Objectives

1. Understand the data and explain the preparation decisions.
2. Define a credible assessment strategy for later purchases.
3. In the homework, select features and assess at least one classifier against
   a simple baseline.
4. Produce one prediction per test ID and reproduce the submitted CSV.

The official Kaggle metric is **F1-Score**, confirmed in the competition
screenshots provided on 19 October 2026. We will compare models using
binary F1 for the positive repeat-purchase class.
We will choose complementary metrics according to the class distribution and
the meaning of false positives and false negatives, rather than relying only
on accuracy. Thresholds and feature choices will be decided using development
data, not test labels or repeated leaderboard feedback.

## Assessment approach
Test purchases occur after the training period. A random stratified split
would mix earlier and later purchases and would not reproduce this setting.

We use a chronological holdout using the latest approximately 20% of
dated training observations. Purchases on the same date stay together. The
90 days immediately before validation are set aside, so every training
target would already be known at the start of validation.

Repeated customers are allowed across periods: the task includes future
purchases by existing customers. Customer identifiers will remain available
for checks but will not be supplied directly to the models.

This is a local development protocol, not a rule explicitly imposed by the
brief. Modeling uses the same chronology inside development. The earlier group notebooks
already explored the full training file, so we should not describe the local
holdout as a completely untouched dataset. Official test labels remain hidden.
